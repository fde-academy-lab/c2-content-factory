## Step 3. Predict, then run: every node's delta (15 minutes)

The chapter's first habit is to read the deltas before any ratio. The picture you just drew splits only
Plus into drivers; the code here computes each tier's change in revenue and splits all three tiers into
the same three drivers: how many customers it had, how often they ordered, and what an order was
worth. Each driver's effect is in Rs lakh, and the three are built so that they add up exactly to
their tier's change.

```python
# PREDICT, TODO 1. Basic gained 4,000 customers and still lost 20 lakh. Which of its drivers pulled
# its revenue down the most?
#   a) "customers"
#   b) "orders per customer"
#   c) "revenue per order"
#   d) "none of them"
basic_driver_prediction = "revenue per order"
print("Your prediction:", basic_driver_prediction)
```

```python
# TODO 2. Which expression gives a tier's delta, its change in revenue from Q1 to Q2?
#   a) kalpa[t]["Q1"]["revenue"] - kalpa[t]["Q2"]["revenue"]
#   b) kalpa[t]["Q2"]["revenue"] - kalpa[t]["Q1"]["revenue"]
#   c) kalpa[t]["Q2"]["revenue"] / kalpa[t]["Q1"]["revenue"]
#   d) kalpa[t]["Q2"]["revenue"] - kalpa["Total"]["Q1"]["revenue"]
delta = {t: kalpa[t]["Q2"]["revenue"] - kalpa[t]["Q1"]["revenue"] for t in (*TIERS, "Total")}
print("Deltas in Rs lakh:", delta)
```

```python
# PREDICT, TODO 3. Student's revenue rose 12 lakh. Which driver carries the rise?
#   a) "revenue per order"
#   b) "orders per customer"
#   c) "all three, about a third each"
#   d) "customers"
student_driver_prediction = "customers"
print("Your prediction:", student_driver_prediction)
```

```python
def driver_effects(q1, q2):
    """Split one tier's change in revenue into three effects, in Rs lakh, that add up to it exactly."""
    opc1, opc2 = q1["orders"] / q1["customers"], q2["orders"] / q2["customers"]
    rpo1, rpo2 = q1["revenue"] / q1["orders"], q2["revenue"] / q2["orders"]
    customers = (q2["customers"] - q1["customers"]) * opc1 * rpo1
    # TODO 4. Which line gives the effect of customers ordering more or less often?
    #   a) q2["customers"] * (opc2 - opc1) * rpo1
    #   b) q1["customers"] * (opc2 - opc1) * rpo2
    #   c) (opc2 - opc1) * rpo1
    #   d) q2["customers"] * opc2 * rpo1
    frequency = q2["customers"] * (opc2 - opc1) * rpo1
    order_value = q2["orders"] * (rpo2 - rpo1)
    return {"customers": customers, "orders per customer": frequency, "revenue per order": order_value}


effects = {t: driver_effects(kalpa[t]["Q1"], kalpa[t]["Q2"]) for t in TIERS}
kit.equation(["the tier's delta", "=", "customers effect", "+", "how often they order", "+", "what an order is worth"],
             title="Three effects, each in Rs lakh, that add up to the tier's delta")
```

```python
def movement(t, driver):
    """A driver's own Q1 and Q2 values, written the way the README shows them."""
    q1, q2 = kalpa[t]["Q1"], kalpa[t]["Q2"]
    if driver == "customers":
        return f"{q1['customers']:,} to {q2['customers']:,}"
    if driver == "orders per customer":
        return f"{q1['orders'] / q1['customers']:.2f} to {q2['orders'] / q2['customers']:.2f}"
    return f"Rs {q1['revenue'] * LAKH / q1['orders']:,.0f} to Rs {q2['revenue'] * LAKH / q2['orders']:,.0f}"


rows = [("Total", "", "804 to 760", f"{delta['Total']:+.1f}")]
for t in TIERS:
    rows.append((t, "", f"{kalpa[t]['Q1']['revenue']} to {kalpa[t]['Q2']['revenue']}", f"{delta[t]:+.1f}"))
    rows += [("", driver, movement(t, driver), f"{effect:+.1f}") for driver, effect in effects[t].items()]
kit.table(["tier", "driver", "Q1 to Q2", "delta, Rs lakh"], rows, caption="Every node of the tree with its delta")

worst = {t: min(effects[t], key=effects[t].get) for t in TIERS}
kit.driver_tree({"label": "Revenue", "note": f"804 to 760, {delta['Total']:+.0f}", "kind": "lit", "children": [
    {"label": t, "note": f"{kalpa[t]['Q1']['revenue']} to {kalpa[t]['Q2']['revenue']}, {delta[t]:+.0f}",
     "kind": "bad" if delta[t] == min(delta[x] for x in TIERS) else "plain",
     "children": [{"label": driver, "note": f"{movement(t, driver)}, {effect:+.1f}",
                   "kind": "bad" if delta[t] == min(delta[x] for x in TIERS) and driver == worst[t] else "plain"}
                  for driver, effect in effects[t].items()]}
    for t in TIERS]}, title="Every node with its delta in Rs lakh: each node's children add up to it", width_per=210)
```

```python
carrier = min(TIERS, key=lambda t: delta[t])
best = {t: max(effects[t], key=effects[t].get) for t in TIERS}
kit.check("your prediction of the driver that pulled Basic down most matches", basic_driver_prediction == worst["Basic"],
          f"you said {basic_driver_prediction!r}; it is {worst['Basic']}, {effects['Basic'][worst['Basic']]:+.1f} lakh")
kit.check("your prediction of the driver behind Student's rise matches", student_driver_prediction == best["Student"],
          f"you said {student_driver_prediction!r}; it is {best['Student']}, {effects['Student'][best['Student']]:+.1f} lakh")
kit.check("Plus carries most of the fall, and inside Plus all 36 lakh is orders per customer, as the chapter's tree says",
          carrier == "Plus" and worst["Plus"] == "orders per customer"
          and abs(effects["Plus"]["orders per customer"] - delta["Plus"]) < 1e-9, f"{carrier}, {worst[carrier]}")
kit.check("the tiers' deltas add up to the total's: -36, -20 and +12 make -44",
          sum(delta[t] for t in TIERS) == delta["Total"] == -44, str(delta))
kit.check("in every tier, the three driver effects add up to the tier's delta",
          all(abs(sum(effects[t].values()) - delta[t]) < 1e-9 for t in TIERS),
          str({t: round(sum(effects[t].values()), 2) for t in TIERS}))
```

**Why, TODO 1:** c holds. Basic's value per order fell from Rs 700 to Rs 625, which across 64,000 Q2
orders is 48 lakh; its orders per customer slipped from 1.50 to 1.45, which is 14 lakh; and its 4,000
new customers added 42 lakh, which is why the net fall is only 20. a pushed Basic's revenue up by 42
lakh; b is real and smaller; d reads the net and misses the parts, since two drivers fell while one
rose.

**Why, TODO 2:** b holds: Q2 minus Q1, so a fall is negative and the tiers add up to the total's -44.
a flips every sign, which still adds up and reads every fall as a rise; c is an index, and indexes do
not add up a tree; d measures each tier against the whole company's Q1.

**Why, TODO 3:** d holds. Student gained 3,000 customers, worth 14.4 lakh at Q1's 1.2 orders a
customer and Rs 400 an order; its orders per customer slipped from 1.20 to 1.12, taking 2.4 lakh
back, and its value per order held at Rs 400. a held flat; b moved down; c spreads the rise over two
drivers that did not carry it.

**Why, TODO 4:** a holds: Q2's customers, the change in orders per customer, and Q1's revenue per
order, which is the order in which the three effects add up exactly. b uses Q1's customers and Q2's
revenue per order, so Basic's effects add up to -17.4 rather than -20; c leaves the customers out, so
the effect shrinks to a fraction of one lakh; d is Q2's orders at Q1's revenue per order, a level
rather than a change.

## Step 4. Trace the running sum (7 minutes)

The tiers' deltas add up to the total one tier at a time. Fill in the trace by hand from the table
first: after each tier, its delta and the running sum, which has to land on the total's -44.

```python
# TODO 5 to 8. Your trace of the loop below, filled in by hand: (tier, its delta, running sum)
my_trace = [
    ("Plus", -36, -36),
    ("Basic", -20, -56),
    ("Student", 12, -44),
]
for row in my_trace:
    print(row)
```

```python
running, real_trace = 0, []
for t in TIERS:
    running += delta[t]
    real_trace.append((t, delta[t], running))
kit.table(["tier", "your delta", "your running sum", "the loop's delta", "the loop's running sum"],
          [(m[0], m[1], m[2], x[1], x[2]) for m, x in zip(my_trace, real_trace)],
          caption="The running sum, tier by tier")
kit.bridge(("Q1 revenue", kalpa["Total"]["Q1"]["revenue"]), [(t, delta[t]) for t in TIERS], end_label="Q2 revenue",
           title="804 to 760 in Rs lakh, tier by tier, on an axis starting at 600", fmt=lambda v: f"{v:,.0f}", lo=600)
```

```python
kit.check("your trace matches the loop after every tier", my_trace == real_trace,
          f"{sum(m == x for m, x in zip(my_trace, real_trace))} of 3 rows match")
kit.check("the running sum lands on the total's delta", real_trace[-1][2] == delta["Total"], str(real_trace[-1]))
```

**Why, the trace:** Basic's delta is 400 minus 420, -20, and the running sum moves from -36 to -56;

## Step 5. Break it on purpose: a ratio before the deltas (10 minutes)

The tempting change: skip the deltas and compute a ratio for every tier first, here the value of an
order, the number a pricing question reaches for. Chapter 5 warns that a ratio on the wrong tier is
precise and useless, and that the wrong answers to Q29 came from this move. Run the ratio-first
version, read what it points at, and read the check.

```python
value_change = {t: 100 * ((kalpa[t]["Q2"]["revenue"] / kalpa[t]["Q2"]["orders"])
                          / (kalpa[t]["Q1"]["revenue"] / kalpa[t]["Q1"]["orders"]) - 1) for t in TIERS}
ratio_first = min(value_change, key=value_change.get)
kit.bars([(t, round(abs(value_change[t]), 1)) for t in TIERS], lit=(TIERS.index(ratio_first),),
         title="The ratio-first view: how far the value of an order fell, percent", fmt=lambda v: f"{v:.1f}")
print(f"Ratio first: {ratio_first}'s order value fell {-value_change[ratio_first]:.1f} percent, "
      f"so {ratio_first} looks like the tier to explain.")
```

**The plausible wrong answer.** Basic's order value fell from Rs 700 to Rs 625, down 10.7 percent,
and no other tier's moved, so the ratio-first reading says Basic carries the fall. That is Q29's
option B, and it points the next quarter at Basic's prices and the monsoon discount while the 36 lakh
in Plus goes unexamined. **Why it is wrong.** The number is precise and belongs to the wrong tier:
Basic's revenue fell only 20 lakh, less than Plus's 36, because Basic also gained 4,000 customers.

```python
kit.check("the check that catches it: the tier the ratio points at is a smaller fall than the largest delta",
          delta[ratio_first] > min(delta[t] for t in TIERS),
          f"{ratio_first} {delta[ratio_first]:+} lakh against Plus {delta['Plus']:+} lakh")
```

**The fix.** Read the deltas, circle the largest, and compute inside that tier only. The tree in
Step 3 already does it: the largest delta is Plus, and inside Plus the driver is orders per customer,
with customers and order value flat.

```python
by_delta = min(TIERS, key=lambda t: delta[t])
kit.check("reading the deltas first points at Plus, and inside Plus at orders per customer",
          by_delta == "Plus" and worst["Plus"] == "orders per customer", f"{by_delta}, then {worst[by_delta]}")
```

### 6.1 Three hypotheses, each against one number in the table (8 minutes)

For each hypothesis, pick the one cell of the table that would move if it were true, then look at it.
The habit is this course's construction, and the self-check asks that every number be one the table
already holds.

```python
# Each hypothesis names the cell that would move if it were true, and the way it would move.
# TODO 9. "Plus customers are downgrading to Basic." Which cell would fall if that were true?
#   a) ("Basic", "customers")
#   b) ("Plus", "orders")
#   c) ("Total", "customers")
#   d) ("Plus", "customers")
# TODO 10. "The customer base moved toward the cheaper tiers." Which cell would rise?
#   a) ("Plus", "revenue")
#   b) ("Student", "revenue")
#   c) ("Basic", "customers")
#   d) ("Total", "orders")
# TODO 11. "Fewer customers placed orders in Q2." Which cell would fall?
#   a) ("Total", "orders")
#   b) ("Total", "customers")
#   c) ("Plus", "customers")
#   d) ("Basic", "customers")
hypotheses = [
    ("Plus customers are downgrading to Basic", ("Plus", "customers"), "falls"),
    ("The customer base moved toward the cheaper tiers", ("Basic", "customers"), "rises"),
    ("Fewer customers placed orders in Q2", ("Total", "customers"), "falls"),
]

verdicts = []
for claim, (tier, field), would in hypotheses:
    q1, q2 = kalpa[tier]["Q1"][field], kalpa[tier]["Q2"][field]
    moved = "rises" if q2 > q1 else "falls" if q2 < q1 else "holds"
    verdict = "supported" if moved == would else "weakened" if moved == "holds" else "contradicted"
    verdicts.append((claim, f"{tier} {field}, {q1:,} to {q2:,}", verdict))
kit.table(["hypothesis", "the one number that tests it", "verdict"], verdicts,
          caption="Three hypotheses, each against a number the table already holds")
```

```python
kit.check("hypothesis 1 is tested by the Plus customer count, which held at 10,000",
          hypotheses[0][1] == ("Plus", "customers"), str(hypotheses[0][1]))
kit.check("hypothesis 2 is tested by the customer count of a cheaper tier, which rose",
          hypotheses[1][1] == ("Basic", "customers"), str(hypotheses[1][1]))
kit.check("hypothesis 3 is tested by the total customer count, which rose from 55,000 to 62,000",
          hypotheses[2][1] == ("Total", "customers"), str(hypotheses[2][1]))
```

**Why, TODO 9:** d holds. Downgrades take customers out of Plus, so the Plus count would fall; it
held at 10,000, which weakens the hypothesis, as Q30 found. a would rise with downgrades and with new
sign-ups alike, so it cannot tell them apart, which was Q30's option A; b tests how often Plus
customers order; c adds up every tier, so it hides where customers went.

**Why, TODO 10:** c holds. If the base moved toward the cheaper tiers, their customer counts rise:
Basic went from 40,000 to 44,000 while Plus held, and Student's 5,000 to 8,000 says the same. a and b
are revenue, which moves with spending as well as with customers; d counts orders, which also rise
when the same customers order more often.

**Why, TODO 11:** b holds. Fewer customers would show as a lower customer count; the total rose from
55,000 to 62,000, which contradicts the claim, as the verdict on Q29's option D says. a counts orders, which can fall
with the same customers ordering less; c and d are single tiers, and the claim is about all of them.

### 6.2 The README, written into a string (8 minutes)

Part one, the tree as a nested list with the delta beside every node, and part two, the hypothesis
table, are generated from the numbers above. Part three is yours: the sentence Meera hears, in your
own words, then the check you would run first. The sentence needs a number from the table, the driver
the tree found and a check, and it should read the way you would say it aloud. Chapter 5 has its own
version in the window-check section; write yours before you look at it again.

```python
# TODO 12. The sentence Meera hears, in your own words, as one string: a number from the table,
# the driver the tree found, and the check you would run before she acts on it.
sentence = ("Revenue fell 44 lakh from Q1 to Q2, and 36 lakh of it is Plus, where the same 10,000 "
            "customers ordered 10 percent less often at the same value per order. Before we act "
            "on it, I will check that both quarters had the same number of trading days and that "
            "no festival fell in only one of them.")

# TODO 13. The check you would run first, as one string: what you would look up, and what would
# change your sentence.
first_check = ("Count the trading days in Q1 and in Q2 and list the festivals that fell in each; if they "
               "differ, compare the same weeks of both quarters before the Plus finding goes to Meera.")


def tree_as_list():
    lines = [f"- Revenue, Rs lakh: 804 to 760, delta {delta['Total']:+.1f}"]
    for t in TIERS:
        lines.append(f"  - {t}: {kalpa[t]['Q1']['revenue']} to {kalpa[t]['Q2']['revenue']}, delta {delta[t]:+.1f}")
        lines += [f"    - {driver}: {movement(t, driver)}, delta {effect:+.1f}"
                  for driver, effect in effects[t].items()]
    return "\n".join(lines)


hypothesis_rows = "\n".join(f"| {claim} | {number} | {verdict} |" for claim, number, verdict in verdicts)
readme = f"""# w00-diagnostic-business

Meera's question, "revenue fell from Q1 to Q2, where did it go?", taken apart on the diagnostic case's table.

## Part one: the decomposition tree

{tree_as_list()}

## Part two: three hypotheses against the table

| Hypothesis | The one number that tests it | Verdict |
| --- | --- | --- |
{hypothesis_rows}

## Part three: the sentence Meera hears

{sentence}

The check to run first: {first_check}
"""
print(readme)
```

### 6.3 The self-check, run against the README (4 minutes)

The guide's three self-check lines, tested on the text itself: the tree's deltas are read back out of
the nested list, and the sentence is searched for a number, a driver and a check. The list of words
you would not say aloud is a short one of analyst's shorthand; read your sentence aloud as the last
test.

```python
def read_tree(text):
    """(depth, delta) for every line of the nested list, read back out of the README's text."""
    nodes = []
    for line in text.splitlines():
        if line.lstrip().startswith("- ") and "delta" in line:
            nodes.append(((len(line) - len(line.lstrip())) // 2, float(line.rsplit("delta", 1)[1])))
    return nodes


def children_add_up(nodes):
    for i, (depth, value) in enumerate(nodes):
        children = []
        for d, v in nodes[i + 1:]:
            if d <= depth:
                break
            if d == depth + 1:
                children.append(v)
        if children and abs(sum(children) - value) > 0.05 * len(children):
            return False
    return True


DRIVER_WORDS = ("less often", "less frequently", "frequency", "orders per customer", "fewer orders")
CHECK_WORDS = ("trading day", "festival", "same weeks", "calendar", "window")
UNSAID = ("AOV", "delta", "QoQ", "YoY", "->", "%", "e.g.", "i.e.", "etc.")

nodes = read_tree(readme)
said = sentence.lower()
has_number = any(ch.isdigit() for ch in sentence)
has_driver = any(w in said for w in DRIVER_WORDS)
has_check = any(w in said for w in CHECK_WORDS)
unsaid = [w for w in UNSAID if w.lower() in said]
table_cells = {(t, f) for t in kalpa for f in ("customers", "orders", "revenue")}

kit.check("self-check 1: in the README's own tree, every node's children add up to it",
          len(nodes) == 13 and children_add_up(nodes), f"{len(nodes)} nodes read back")
kit.check("self-check 2: every hypothesis names a number the table already holds",
          all(cell in table_cells for _, cell, _ in hypotheses), str([cell for _, cell, _ in hypotheses]))
kit.check("self-check 3: the sentence carries a number, a driver and a check",
          has_number and has_driver and has_check, f"number {has_number}, driver {has_driver}, check {has_check}")
kit.check("self-check 3, continued: none of the shorthand you would not say aloud",
          not unsaid, "none found" if not unsaid else f"found {unsaid}")
kit.check("the check to run first names the window it looks at",
          any(w in first_check.lower() for w in CHECK_WORDS), first_check[:70])
```
