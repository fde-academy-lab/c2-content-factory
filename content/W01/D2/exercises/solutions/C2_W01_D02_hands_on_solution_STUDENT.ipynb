# Solution: the same drop, decomposed a different way

The session decomposed the fall by **segment** and found Retail-Plus. Marketing's reply is short:

> "Segment is your cut, not mine. I buy media by channel. Show me the same thing by channel, and
> tell me whether it says the same thing."

Every blank is filled and the notebook is executed, so the numbers below are the ones the file
actually holds.

```python
import pathlib
import sys

root = next(p for p in pathlib.Path.cwd().resolve().parents if (p / "scripts" / "c2kit.py").exists())
sys.path.insert(0, str(root / "scripts"))
import c2kit as kit

ORDERS = kit.load_records("C2_W01_D02_orders_STUDENT.py")
q1 = [o for o in ORDERS if o["quarter"] == "Q1"]
q2 = [o for o in ORDERS if o["quarter"] == "Q2"]
print(f"{len(q1)} rows in Q1 and {len(q2)} in Q2")
```

## 2. Write the describe function again, from memory

It has to **return** rather than print, because section 3 calls it once per channel.

```python
def describe(rows):
    people = {o["customer_id"] for o in rows}
    total = sum(int(o["amount"]) for o in rows)
    return {"orders": len(rows), "customers": len(people),
            "per_customer": len(rows) / len(people)}

check_rows = describe(q1)
print(check_rows)
```

```python
kit.check("describe returns a dictionary", isinstance(check_rows, dict),
          f"got {type(check_rows).__name__}")
kit.check("Q1 has 114 rows", check_rows["orders"] == 114, f"got {check_rows['orders']}")
kit.check("Q1 has 69 distinct customers", check_rows["customers"] == 69,
          f"got {check_rows['customers']}")
```

## 3. Orders per customer, by channel

One row per channel, both quarters, and the change between them.

```python
channels = sorted({o["channel"] for o in ORDERS})
rows, moves = [], {}
for ch in channels:
    a = describe([o for o in q1 if o["channel"] == ch])
    b = describe([o for o in q2 if o["channel"] == ch])
    change = 100 * (b["per_customer"] / a["per_customer"] - 1)
    moves[ch] = change
    rows.append((ch, a["orders"], b["orders"], f"{a['per_customer']:.2f}",
                 f"{b['per_customer']:.2f}", f"{change:+.1f}%"))
kit.table(["channel", "Q1 orders", "Q2 orders", "Q1 per customer", "Q2 per customer", "change"],
          rows, caption="The same fall, cut by channel")
```

```python
kit.check("three channels found", len(channels) == 3, f"got {channels}")
kit.check("one channel is clearly the worst",
          min(moves.values()) < -15, f"worst is {min(moves.values()):+.1f} percent")
```

## 4. Now cut it by city, and be careful

Same code, one word changed. Then look at how many customers sit behind each number before you
believe any of it.

```python
cities = sorted({o["city"] for o in ORDERS})
rows = []
for c in cities:
    a = describe([o for o in q1 if o["city"] == c])
    b = describe([o for o in q2 if o["city"] == c])
    rows.append((c, a["customers"], b["customers"], f"{a['per_customer']:.2f}",
                 f"{b['per_customer']:.2f}",
                 f"{100 * (b['per_customer'] / a['per_customer'] - 1):+.1f}%"))
kit.table(["city", "Q1 customers", "Q2 customers", "Q1 per customer", "Q2 per customer", "change"],
          rows, caption="The same fall again, cut by city")
```

```python
smallest = 6   # Hyderabad moved -44 percent on six customers in Q2
kit.check("you looked at the base behind the biggest move", smallest < 10,
          f"{smallest} customers")
```

## 7. Two sentences

One for Marketing, in channel language. One for the head of Retail-Plus, in segment language.
Both have to be true at the same time, and neither may state a cause as a fact.

> **To Marketing:** "Web orders per customer fell 22 percent between the quarters while app was
> flat, so the fall sits in web rather than across your media. Most of the web fall is Retail-Plus
> members, so a media change alone is unlikely to recover it."

> **To Retail-Plus:** "Your tier is down 49 percent on orders per member, against 5 percent for
> Retail-Core, and it shows up most in the web channel. The reorder-button complaint is a
> plausible cause and I have not tested it; reorder events per member either side of the six
> weeks, against Retail-Core, would settle it."

## Why each letter

| | Key | Why the others fail |
|---|---|---|
| **Q1** | `b` | Neither cut is wrong. A Retail-Plus member ordering on web sits in both, so the two findings are the same customers described twice. Reporting only one of them is how a team ends up briefing two departments on contradictory stories. |
| **Q2** | `c` | Hyderabad moved 44 percent on six customers in Q2. The size of a percentage says nothing about how much of it is noise; the base behind it does. Thursday turns this instinct into a rule. |
| **Q3** | `c` | Marketing acts in channel language, so the cut has to reach them in it. Saying which segment sits inside the channel is what stops them buying more web media to fix a broken reorder button. |
| **Q4** | `c` | `print` returns `None`, so `a["per_customer"]` raises `TypeError: 'NoneType' object is not subscriptable` on the first channel. A function that prints cannot be built on, which is the whole reason `describe` returns. |
| **Q5** | `b` | Web is where the fall is. Why is a fact about a broken feature, a price change or a competitor, and none of those is in this file. Naming a location as a cause is the single most common way a good analysis goes wrong in front of a stakeholder. |
