## Setup

The same 44 profiled orders. The summary below was written by somebody in a hurry and it raises nothing at all.

````python
import statistics

orders = kit.load_csv("C2_W01_D04_profiled_STUDENT.csv")
for r in orders:
    r["amount"] = int(r["amount"])

print(len(orders), "profiled orders read from ../data/")
print(sorted({r["segment"] for r in orders}))
````

## Step 1. Read the broken table before you touch it

It runs. It prints a tidy table. Decide what is wrong with it before you change a line.

````python
kit.flow(["read the broken table", "fix the statistic", "restore the denominator", "state the floor"], lit=0)
````

````python
# TODO 1. What is wrong with the typical column?
#   a) it is the median, which ignores size
#   b) it is the mean, which one order owns
#   c) it is the mode, which repeats nothing
#   d) nothing, since the arithmetic is right
broken = {}
for r in orders:
    key = r["segment"]
    b = broken.setdefault(key, {"amounts": [], "returned": 0, "count": 0})
    b["amounts"].append(r["amount"])
    b["count"] += 1
    if r["status"] == "returned":
        b["returned"] += 1

for key in sorted(broken):
    b = broken[key]
    print(f"{key:<12} typical Rs {sum(b['amounts']) / b['count']:>9,.0f}   "
          f"rate {100 * b['returned'] / b['count']:>5.1f}%")

first_problem = "the mean, which KR4232 owns"
print("first problem:", first_problem)
````

````python
kit.check("Retail-Core's broken typical is above Rs 30,000",
          sum(broken["Retail-Core"]["amounts"]) / broken["Retail-Core"]["count"] > 30000)
kit.check("no actual Retail-Core order is anywhere near it",
          sorted(broken["Retail-Core"]["amounts"])[-2] < 5000)
kit.check("you named the mean", "mean" in first_problem)
````

It is the mean rather than the median. The mode repeats almost nothing on a money field. Saying nothing is wrong because the arithmetic is right is the exact belief the day exists to break: the number is correct and the description is not.

## Step 2. Fix the statistic

Swap in the statistic that survives one enormous order, and watch three of the four rows barely move while one collapses.

````python
kit.flow(["read the broken table", "fix the statistic", "restore the denominator", "state the floor"], lit=1)
````

````python
# TODO 2. What replaces the mean?
#   a) max(b["amounts"])
#   b) sum(b["amounts"])
#   c) statistics.median(b["amounts"])
#   d) sum(b["amounts"]) / len(b["amounts"])
fixed = {}
for key, b in broken.items():
    fixed[key] = {"typical": statistics.median(b["amounts"]), "count": b["count"], "returned": b["returned"]}

for key in sorted(fixed):
    print(f"{key:<12} typical Rs {fixed[key]['typical']:>7,.0f}")
````

````python
kit.check("Retail-Core's typical order is now about Rs 1,910",
          abs(fixed["Retail-Core"]["typical"] - 1910) < 50)
kit.check("every segment's typical order is now under Rs 2,100",
          all(s["typical"] < 2100 for s in fixed.values()))
kit.check("the counts did not change", sum(s["count"] for s in fixed.values()) == 44)
````

The maximum and the sum each describe something other than a typical order. The last option is the mean written out longhand, which is the line you are replacing.

## Step 3. Restore the denominator

A rate with no count beside it cannot be argued with. Put the count back and the ranking reads differently.

````python
kit.flow(["read the broken table", "fix the statistic", "restore the denominator", "state the floor"], lit=2)
````

````python
# TODO 3. What does each line have to carry?
#   a) rate_alone
#   b) with_name
#   c) with_median
#   d) with_count, the count the rate rests on
for key in sorted(fixed, key=lambda k: fixed[k]["returned"] / fixed[k]["count"]):
    s = fixed[key]
    pct = format(100 * s["returned"] / s["count"], ".1f")
    rate_alone = key.ljust(12) + " " + pct + "%"
    with_name = key.ljust(12) + " " + pct + "% for " + key
    with_median = key.ljust(12) + " " + pct + "%  median Rs " + format(s["typical"], ",.0f")
    with_count = key.ljust(12) + " " + pct + "%  on " + str(s["count"]) + " orders"
    line = with_count
    print(line)
````

````python
kit.check("the line carries the count", "orders" in line)
kit.check("the best-ranked segment has nine orders behind it",
          min(fixed.values(), key=lambda s: s["returned"] / s["count"])["count"] == 9)
````

The rate alone is the version that travels into a slide and cannot be questioned. The name is a label rather than evidence. The median answers a different question from the rate.

## Step 4. State the floor, and say what it means here

A threshold you are prepared to defend, not a law of nature. Then say what it means that every segment in this file trips it.

````python
kit.flow(["read the broken table", "fix the statistic", "restore the denominator", "state the floor"], lit=3)
````

````python
# TODO 4. What is the finding?
#   a) the segments should be merged
#   b) the floor is too high
#   c) the file is too small to rank segments at all
#   d) the rates should be rounded
TRUST_FLOOR = 30
below = [k for k, s in fixed.items() if s["count"] < TRUST_FLOOR]

finding = "a 44 order file cannot rank segments, and that is the finding"
print(len(below), "of", len(fixed), "segments are below the floor")
print(finding)
````

````python
kit.check("every one of the four segments is below the floor", len(below) == 4)
kit.check("your finding says the file is too small", "cannot rank" in finding)
kit.check("and the largest segment is still only fourteen orders",
          max(s["count"] for s in fixed.values()) == 14)
````
