## Section 1: the guided build

The question this time is about size rather than status. How many of these thirty orders are above Rs 2,000, and what do they come to? A threshold question like that one turns up wherever a rule has a cut-off in it.

Build it with the trainer, one step at a time, in your own Codespace. Start with a single record, so that you can watch the condition work before it runs thirty times.

````python
r = records[1]

print(r["order_id"])
print(int(r["amount"]))
print(int(r["amount"]) > 2000)
````

The condition hands back `True` or `False` for one record. Put it inside a loop and the same question is asked of all thirty, and a counter set above the loop remembers how many times the answer came back `True`.

````python
count = 0
for r in records:
    if int(r["amount"]) > 2000:
        count = count + 1

print(count)
````

Thirteen orders clear the threshold. A count says nothing about what those orders are worth, so add the second accumulator to the same walk of the records rather than writing a second loop.

````python
count = 0
total = 0
for r in records:
    if int(r["amount"]) > 2000:
        count = count + 1
        total = total + int(r["amount"])

print(count)
print(total)
````

````python
kit.check("thirteen orders are above Rs 2,000", count == 13, f"count is {count}")
kit.check("they come to Rs 35,020", total == 35020, f"total is {total}")
kit.check("this is a different thirteen from yesterday's delivered thirteen",
          {r["order_id"] for r in records if int(r["amount"]) > 2000}
          != {r["order_id"] for r in records if r["status"] == "delivered"})
````

Say that answer the way you would say it in the room. Thirteen of the thirty orders are above Rs 2,000, and those thirteen come to Rs 35,020.

One warning about that thirteen. The first notebook also ended on a thirteen, which was the count of delivered orders, and these two thirteens count different records. One condition asked about status and the other asked about amount, and the two counts landing on the same number is a coincidence of this file. When you report a number, say which condition produced it, because the number on its own cannot tell anybody that.

### Milestone: where this shows up in production

A threshold is how a business rule gets written down. Free delivery above a value, a second approval above a value, a review queue for anything above a value: each of those is a condition sitting in somebody's code, and each of them produces a count and a total that somebody reports upwards.

The work you just did is the whole of that job. The care is in the condition, and the argument in the room is about where the threshold sits and what happens to the records that could not be compared at all.

## Section 2: a list is an order you can trust

`records` is a list, so the thirty orders sit in a fixed order and each one has a position. You reach a position with square brackets, counting from zero, so the first order is `records[0]` and the last of the thirty is `records[29]`.

A negative index counts back from the end. A slice takes a run of positions and stops before the second number, so `records[10:13]` hands you three orders.

````python
print(records[0]["order_id"])
print(records[29]["order_id"])
print(records[-1]["order_id"])

print()
for r in records[10:13]:
    print("in the slice:", r["order_id"])
````

`records[29]` and `records[-1]` named the same order, which is the point of the negative index. You can take the last one without knowing how many there are, and the code keeps working when the file grows.

### Building a list of your own

`append` puts one item on the end of a list. Start with an empty list, walk the records, and keep the ones that qualify.

````python
large_ids = []
for r in records:
    if int(r["amount"]) > 2000:
        large_ids.append(r["order_id"])

print(len(large_ids))
print(large_ids)
````

````python
kit.check("the list of names is as long as the count was", len(large_ids) == 13)
kit.check("every kept name is a real order id", set(large_ids) <= {r["order_id"] for r in records})
````

That is the condition from section 1, keeping the names of the orders instead of counting them, and the length of the list agrees with the count you already have. Keeping the ids is what lets somebody check your number one record at a time, which is the difference between a number and a number you can defend.

### The trap: b = a

An assignment gives a value a name. When the value is already a list, the assignment gives that one list a second name, and both names reach the same list.

````python
a = [1, 2, 3]
b = a
b.append(9)

print("a is", a)
print("b is", b)
print("len(a) is", len(a))
````

`a` grew, and no line of that code mentions `a` after the second one. There is one list here wearing two names, and appending through either name changes the thing both names reach.

This is the most common surprise in a first week of Python, and it is worth ten seconds of your attention now to save you an afternoon later in the programme.

### Copying on purpose

When you want a second list, say so. `list(a)` builds a new list holding the same items. A full slice, `a[:]`, does the same thing, and you will meet both forms in other people's code.

````python
a = [1, 2, 3]
b = list(a)
b.append(9)

print("a is", a)
print("b is", b)

c = a[:]
c.append(7)

print("a is still", a)
print("c is", c)
````

````python
kit.check("a was left alone once the copy was made on purpose", a == [1, 2, 3], f"a is {a}")
kit.check("both copies grew independently", b == [1, 2, 3, 9] and c == [1, 2, 3, 7])
````

### What section 2 established

### Milestone: where this shows up in production

A list handed to one piece of code and quietly changed by another is a bug that survives review, because the line doing the damage looks harmless and sits a long way from the line that reads the damage. The habit that prevents it is copying at the boundary: code that is going to change a list works on its own copy and hands that copy back.

You meet the same rule tomorrow, one level up, where a record gets copied before anything cleans it.

## Section 3: a record is a dictionary

Each order in that list is a dictionary. A dictionary holds pairs, each pair a name and a value, and you fetch a value by writing its name in square brackets.

`records[15]` is one of the two orders in this file that carries a discount, so it has one more pair in it than its neighbours do.

````python
order = records[15]

print(order)
print(order["order_id"])
print(order["segment"])
print(order["amount"])
print(order["discount"])
````

A list answers the question "give me the sixteenth thing". A dictionary answers the question "give me the thing called `segment`". Both are useful and they answer different questions.

Position is fragile. Add a column to tomorrow's export and every position moves along by one, while every name stays exactly where it was. That is why a record is fetched by field name in every artifact in this programme.

### The deliberate failure: a name that is not there

That order carries a `discount` and it is one of only two in the file that do. Ask any of the other twenty-eight orders for a discount and the dictionary has nothing to hand you.

The `try` and `except` wrapper below is the same plumbing you saw in the first notebook, and it is here so this notebook keeps running top to bottom. You meet `try` and `except` properly tomorrow.

````python
try:
    print(records[0]["discount"])
except KeyError as e:
    print("KeyError:", e)
````

````python
absent = [r["order_id"] for r in records if "discount" not in r]
kit.check("the KeyError is real, since 28 of the 30 orders carry no discount", len(absent) == 28,
          f"{len(absent)} orders have no discount key")
kit.check("KR4200 is one of them", "KR4200" in absent)
````

Run that line in a cell of your own, with no wrapper around it, and this is the screen:

```
Traceback (most recent call last):
  Cell In[N], line 1
KeyError: 'discount'
```

`N` is whatever number your own cell carries. The last line names the key it went looking for, and the quotes around it tell you that it was looking for a name rather than a position.

Read the message precisely, because the wording is doing real work. The field is absent from that record. Absent is a different thing from empty, and it is a different thing again from zero. KR4200 was never given a discount, and the file says so by leaving the name out altogether.

### Asking with a default

`.get()` asks for a name and takes a default for the case where the name is missing, so your code carries on and you are the one who decides what missing means.

````python
print(records[0].get("discount", 0))
print(records[15].get("discount", 0))
````

The default you write is a decision about the business, and it belongs in your notes next to the number it produced. Watch what two different defaults do to the same thirty records.

````python
discount_with_zero = 0
for r in records:
    discount_with_zero = discount_with_zero + r.get("discount", 0)

discount_with_hundred = 0
for r in records:
    discount_with_hundred = discount_with_hundred + r.get("discount", 100)

print(discount_with_zero)
print(discount_with_hundred)
````

````python
kit.check("a default of zero reports what the file actually records",
          discount_with_zero == 250, f"Rs {discount_with_zero}")
kit.check("a default of a hundred invents Rs 2,800 that nobody gave",
          discount_with_hundred - discount_with_zero == 2800,
          f"Rs {discount_with_hundred} against Rs {discount_with_zero}")
kit.check("only two orders carry a discount at all",
          len([r for r in records if "discount" in r]) == 2)
````

### The choice, drawn

````python
kit.tree(
    {"label": "a record has no discount key",
     "branches": [
         ("default 0", {"label": "Rs 250 reported, which is what the file holds"}),
         ("default 100", {"label": "Rs 3,050 reported, and Rs 2,800 of it was invented"}),
         ("no default", {"label": "KeyError, and the loop stops on the first record"})]},
    taken=["default 0"], title="what .get() decides for you")
````

Rs 250 is what the file actually records, which is KR4215 at Rs 75 and KR4223 at Rs 175 and nothing anywhere else. Rs 3,050 is what a default of 100 invents on the twenty-eight orders that never had a discount at all.

Both cells ran cleanly and only one of them is an answer. The default is yours, you own it, and you say it out loud when you hand the number over.

## Section 4: the shape everything else takes

`records` is a list of dictionaries, and that is what a dataset is for the rest of this programme. Whatever the file looks like tomorrow, this is the shape it takes once it is in memory.

Every question left in the day is the same few lines with a different condition. Here is the shape worked once on a segment, so that you see it whole before you write your own.

````python
count = 0
total = 0
for r in records:
    if r["segment"] == "Retail-Plus":
        count = count + 1
        total = total + int(r["amount"])

print(count)
print(total)
````

Eight Retail-Plus orders, coming to Rs 14,050. One walk of the records, one condition, two accumulators, and one sentence at the end that names the condition.
