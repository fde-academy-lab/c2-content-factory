## Section 1: the workbench

`**[open the workbench]** > **[the kernel remembers]** > [type decides] > [walk the records] > [the business answer]`

The kernel is the Python that runs your cells, and it is a workbench that keeps whatever you put on it for as long as it is running. Running a cell is you putting something on the bench.

The bench already holds `records`, because the setup cell ran. It holds everything else you have defined since the kernel started, and it holds nothing you have only typed.

````python
bench_note = "the kernel is holding this string for you"
runs_since_reset = 0

print(bench_note)
print("orders on the bench:", len(records))
print("runs recorded since the marker was set:", runs_since_reset)
````

### Cells run in the order you run them

The notebook is a stack of cells on a screen and the kernel has no opinion at all about that stack. The kernel sees the order you clicked run in, so if you ran the third cell first then the third cell ran first.

The number in square brackets to the left of a cell is the execution counter, and it counts the runs the kernel has done. Read the counters when an output surprises you, because the counters are the truth and the positions on screen are only a layout.

The cell below makes that visible. Run it three times in a row and watch the number climb. Then run the marker cell above once and run it again.

````python
runs_since_reset = runs_since_reset + 1

print("runs recorded since the marker was set:", runs_since_reset)
````

Nothing on disk changed while that number climbed. The count lives on the bench and nowhere else, and running the marker cell again puts it back to zero.

That is the whole idea of kernel state. The kernel holds exactly what you gave it, in the order you gave it, and it forgets all of it the moment you restart.

### The deliberate failure: the name comes off the bench

Run a counting cell before the setup cell and the name `records` was never put on the bench, so the kernel has nothing to hand your loop. That is the first error most people meet in a notebook, and the cell below stages it on demand.

Two pieces of plumbing sit in that cell and neither one is today's lesson. `del` takes a name off the workbench, and it is a prop rather than a lesson: it is here only so the failure can happen when we want it. The `try` and `except` wrapper catches the failure and prints it, so this notebook keeps running top to bottom. You meet `try` and `except` properly tomorrow.

````python
records_backup = records     # a prop, so we can put the list back afterwards
del records                  # take the name off the workbench

try:
    print(len(records))
except NameError as e:
    print("NameError:", e)

records = records_backup     # put it back
print("records is on the bench again, holding", len(records), "orders")
````

The cell above put `records` back on the bench, so that line prints thirty for you now. On a kernel where the setup cell has not run, the same line sits alone in a cell with no wrapper around it and this is what the screen gives you:

```
Traceback (most recent call last):
  Cell In[N], line 1
NameError: name 'records' is not defined
```

`N` is whatever number your own cell carries, so ignore it and read the last line. The error names the exact word it went looking for, which is `records`.

### The recovery drill

Restart the kernel, then run all cells from the top. You are back where you were in a few seconds, because the file on disk never changed and the cells put the bench back the way it was.

Do the drill now, on purpose, while nothing is at stake. You will do it many times this week, and the people who are calm about it are the people who have already done it once by choice.

## Section 2: what a value is

`[open the workbench] > [the kernel remembers] > **[type decides]** > [walk the records] > [the business answer]`

Every value in Python has a type, and the type decides what an operator means. Four types carry today.

`str` holds text, which is what an order id and a status are. `int` holds a whole number, which is what an amount should be. `float` holds a number with a decimal part. `bool` holds `True` or `False`, and every comparison you write hands one back.

````python
order_id = "KR4224"
amount = 1460
amount_with_paise = 1460.5
is_large = amount > 1000

print(type(order_id))
print(type(amount))
print(type(amount_with_paise))
print(type(is_large))
````

`type()` is the question you ask a value when an operator behaves in a way you did not expect. Ask it before you change any code, because the answer usually ends the investigation.

Now ask it of the file's own values instead of ones you typed yourself.

````python
print(records[0]["order_id"], type(records[0]["amount"]).__name__)
print(records[1]["order_id"], type(records[1]["amount"]).__name__)

print()
for r in records[:5]:
    print(r["order_id"], r["amount"], type(r["amount"]).__name__)
````

Read the first column of that output next to the last one. KR4200 stores its amount as the text `"4500"` and every other record in the file stores a number.

Three facts about KR4200 matter today. Its amount is text, it is the largest amount in the file, and it is a returned order. All three of them turn up before the day is out.

Nobody typed that record wrongly on purpose. A field arrives as text because it passed through an export, a spreadsheet or another team's system, and that is most of the data you will ever be handed.

## Section 3: the comparison Python refuses

Money questions start with a comparison. Ask for the orders above Rs 2,000 and the loop walks into KR4200 on its very first turn.

The `try` and `except` wrapper below is the same plumbing as before, so the notebook survives the failure and you can read the message at your own pace.

````python
count = 0
try:
    for r in records:
        if r["amount"] > 2000:
            count = count + 1
except TypeError as e:
    print("TypeError:", e)
````

Run those lines in a cell of your own, without the wrapper, and this is the screen:

```
Traceback (most recent call last):
  Cell In[N], line 3
TypeError: '>' not supported between instances of 'str' and 'int'
```

`N` is whatever number your own cell carries. Read the last line first, every time, and ask three questions in order.

1. What is the exception type? It is `TypeError`, so the problem is what the values are.
2. Which line is mine? Line 3, the comparison.
3. What was it holding? A `str` and an `int`, in that order, which names KR4200 for you before you go looking.

Python was asked whether a piece of text is larger than a number, and it stopped. A spreadsheet would place that text somewhere in the sort order and show you a total with no warning attached to it. You lose ten seconds to this error and you keep the truth, which is the trade you want every time.

### The fix, at the point of use

`int()` takes a value and hands back the whole number it stands for. Convert where you use the value, so the record itself stays exactly as it arrived and anyone can still see what the source sent you.

````python
print(records[0]["amount"], type(records[0]["amount"]).__name__)
print(int(records[0]["amount"]), type(int(records[0]["amount"])).__name__)
print(int(records[0]["amount"]) > 2000)

print()
for r in records[:5]:
    print(r["order_id"], int(r["amount"]) > 2000)
````

The comparison runs on every record now, because both sides are numbers by the time the operator sees them. The record still holds its text, which is what you want: your code decided what to do with the value and the file kept its evidence.

Every conversion in this notebook works because that text happens to hold digits. Tomorrow that assumption gets tested properly, and you learn what to do on the day it fails.

## Section 4: walk the records and come back with a number

`[open the workbench] > [the kernel remembers] > [type decides] > **[walk the records]** > [the business answer]`

A dataset is a stack of cards and a loop deals you one card at a time. The loop below calls each card `r`, and everything indented under it happens once per card.

````python
for r in records[:3]:
    print(r["order_id"], r["status"], r["amount"])
````

Thirty records means thirty turns through the indented lines. The way a field is fetched by its name gets its proper treatment in the next notebook.

### Choosing between cases

`if` runs its block when the condition is `True`. `elif` is asked only when everything above it was `False`. `else` catches whatever is left. Python takes the first branch that holds and never looks at the rest of them.

````python
amount = int(records[24]["amount"])

if amount > 2000:
    band = "large"
elif amount > 1500:
    band = "medium"
else:
    band = "small"

print(records[24]["order_id"], amount, band)
````

### The count accumulator

An accumulator is a name you set once, above the loop, and change on every turn that qualifies. Counting is the smallest version of it, which is to start at zero and add one.

````python
count = 0
for r in records:
    if r["status"] == "delivered":
        count = count + 1

print(count)
````

Thirteen orders were delivered. Totalling them is the same shape with the amount going in where the one was.

````python
total = 0
for r in records:
    if r["status"] == "delivered":
        total = total + int(r["amount"])

print(total)
````

Rs 25,720, which is the number the notebook handed you at the start of the day. You have now written every line that produced it.

The two accumulators go in one loop when you want both numbers, and then a single walk of the records answers the whole question.

````python
count = 0
total = 0
for r in records:
    if r["status"] == "delivered":
        count = count + 1
        total = total + int(r["amount"])

print(count)
print(total)
````

### Find the mistake

The cell below is the same work with one line in the wrong place. It does not crash. It prints a number, the number is wrong, and nothing on the screen says so.

Read it before you run it and decide what you expect to see.

````python
for r in records:
    if r["status"] == "delivered":
        total = 0
        total = total + int(r["amount"])

print(total)
````

Rs 1,460 is the amount of KR4224, the last delivered order in the file, and it is a total of nothing at all. The accumulator was set back to zero on every qualifying record, so the loop kept only the last record it saw and threw the other twelve away.

The rule that prevents it is one line long. The accumulator is set to zero once, above the loop, where the loop cannot reach it.

Learn this failure by its shape. When a total comes back far too small and it happens to look like a value you could find in the data, go and find the record that explains it before you touch anything else.

### Milestone: where this shows up in production

Every daily business number you have ever been shown is an accumulator. A revenue tile on a dashboard is a loop, a condition and a running total, executed on a schedule against yesterday's records, and the only reason anyone can trust it is that somebody can point at the condition and say what it includes.

The three things that go wrong in production are the three you have now seen on your own screen. The condition selects the wrong records, the accumulator is reset where it should not be, and a value's type quietly changes what the comparison means.
