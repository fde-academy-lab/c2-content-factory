## 1. The kernel remembers what you ran, not what you see

The page shows cells in order, top to bottom. The kernel, the Python process behind the page, only
knows the cells you actually ran, in the order you ran them, since it last restarted. A name exists
for the kernel only after the cell that makes it has run.

**Predict before you run.** The next cell uses a name, `DELIVERED`, that no cell above has made.
What does it print: a) an empty list; b) 0; c) an error naming `DELIVERED`; d) nothing at all?

The `with kit.expect_error()` line catches the error so the rest of the notebook keeps running. On
your own screen, without it, Jupyter prints the same last line under a longer traceback.

````python
with kit.expect_error() as missing:
    print(len(DELIVERED))      # no cell above has made DELIVERED
````

````python
kit.side_by_side(
    kit.flow(["cell 1\nloads ORDERS", "cell 2\ncounts orders", "cell 3\nprints the count"],
             title="What the page shows, top to bottom", show=False),
    kit.flow(["fresh kernel\nknows no names", "cell 3, run first\nlen(ORDERS)",
              "NameError\nORDERS was never made"],
             kinds=[None, None, "bad"], title="What the kernel ran, since its restart", show=False),
)
````

**What happened.** The answer is c: `NameError: name 'DELIVERED' is not defined`. The same thing
happens in the classroom when a learner restarts the kernel and runs the third cell first, because
the page above the cell is not the kernel's memory.

The opposite failure is worse because it is silent: a notebook that only works because a cell you
since deleted ran earlier. The only test that catches it is the one to run before every commit,
every share and every answer that goes to Meera.

| Habit | What it means | What it catches |
|---|---|---|
| Restart and Run All | Every cell runs once, top to bottom, from an empty kernel | Hidden state from cells run out of order or deleted |
| Top to bottom | Cells written in the order they must run | A cell that leans on one further down |
| Read the brackets | `In [1]`, `In [2]`, `In [3]` down the page | Gaps and jumps that show the run order was not the page order |

````python
kit.check("a name no cell has made raises NameError", missing.name == "NameError",
          f"{missing.name}: {missing.message}")
kit.check("the error names the missing name", "DELIVERED" in missing.message, missing.message)
````

Week 0's librarian's log stored one lent book as one dictionary. A Kalpa order is the same idea
with different fields: seven named values, reached by name, never by position. That is what lets a
question ask for `order["amount"]` without caring where the amount sits in the record.

**Predict before you run.** What is the type of the value stored under `"amount"` in the first
order, and what is the type of the value under `"order_date"`?

````python
first = ORDERS[0]
kit.table(["key", "value", "type", "the question it answers"],
          [("order_id", first["order_id"], type(first["order_id"]).__name__, "which order: one row each"),
           ("customer_id", first["customer_id"], type(first["customer_id"]).__name__, "who bought, which is how customers are counted"),
           ("segment", first["segment"], type(first["segment"]).__name__, "which kind of customer, which Tuesday groups by"),
           ("channel", first["channel"], type(first["channel"]).__name__, "where they bought: app, web or store"),
           ("order_date", first["order_date"], type(first["order_date"]).__name__, "when, as text for now"),
           ("amount", first["amount"], type(first["amount"]).__name__, "for how much, which is how revenue is summed"),
           ("status", first["status"], type(first["status"]).__name__, "whether it stayed sold")],
          caption="ORDERS[0], one Kalpa order, field by field")
````

## 4. Counting is an accumulator

An accumulator makes three moves, the same three the librarian's running total made in Week 0:
**start** a total before the loop, **update** it once per record inside the loop, and **finish** by
reading it after the loop. Every leaf on the tree is one of these three moves with a different
update.

The cell below counts the orders and keeps a desk check of the first five passes, which is how a
loop is read by hand before it is trusted.

````python
count = 0
passes = []
for order in ORDERS:
    count = count + 1
    if count <= 5:
        passes.append((count, order["order_id"], f"count is now {count}"))
print(count)

kit.table(["pass", "the order it visited", "after the update"], passes,
          caption="A desk check: the first five passes of the loop")
kit.flow(["start\ncount = 0, before the loop", "update\ncount = count + 1, once per order",
          "finish\nprint(count), after the loop"], lit=1,
         title="Three moves, and two classic bugs: the start inside the loop, the print inside it")
````

**Two bugs worth seeing once.** Put `count = 0` inside the loop and it resets on every pass, so the
count ends at 1. Put `print(count)` inside the loop and it prints thirty times. Both run without an
error, which is why the desk check exists: a wrong answer that does not crash is the expensive
kind.

### Predict before you run: the revenue loop

The same three moves, summing each order's amount instead of adding one. What does the next cell
print?

a) 544810; b) thirty lines, one running total per order; c) an error part way through the list;
d) 0, because revenue starts at zero.

````python
with kit.expect_error() as stopped:
    revenue = 0
    for order in ORDERS:
        revenue += order["amount"]
    print(revenue)
````

**What happened.** The answer is c. This is exactly what Jupyter prints on a screen without the
`with` line, and it is read from the bottom up:

```text
TypeError                                 Traceback (most recent call last)
Cell In[2], line 3
      1 revenue = 0
      2 for order in ORDERS:
----> 3     revenue += order["amount"]
      4 print(revenue)

TypeError: unsupported operand type(s) for +=: 'int' and 'str'
```

| Read | What it says here | What it tells you |
|---|---|---|
| The last line, first | `TypeError` | The kind of problem: an operation met a type it cannot use |
| The operation | `+=` | Adding to the running total is what failed |
| The two sides | `'int' and 'str'` | A number on the left, text on the right |
| The arrow | line 3 | Where it gave up; the lines around it are context |

## 5. The fix for today, and what it costs

`int()` turns text that looks like a whole number into a number, so the loop can add every amount.
It is a patch for today: the total becomes right, and the question of why an amount arrived as
text, and whether a bigger file holds more of them, stays open until Wednesday.

````python
revenue = 0
for order in ORDERS:
    revenue += int(order["amount"])
print(revenue)
print(kit.rupees(revenue))
````

````python
kit.check("the loop stopped on a type error", stopped.name == "TypeError", stopped.message)
kit.check("the error came from adding text to a number",
          "'int' and 'str'" in stopped.message, stopped.message)
kit.check("thirty orders were counted", count == 30, f"got {count}")
kit.check("with int(), revenue is Rs 5,44,810", revenue == 544810, kit.rupees(revenue))
kit.check("revenue matches notebook 1's booked total", revenue == 544810 and len(ORDERS) == 30)
````

### Depth: where the patch breaks

The same fix fails the moment the text carries an Indian comma, which is how a person types one
lakh twenty thousand.

````python
with kit.expect_error() as comma:
    int("1,20,000")
````

A conversion that works on today's value is not a cleaning strategy. Cleaning means deciding, per
field, what counts as a valid value and what happens to the rows that are not, which is
Wednesday's work. The honest sentence to Meera today is: "Revenue, all booked orders, is Rs
5,44,810; one amount arrived as text and was converted, and we are checking the file for others."

````python
kit.check("int() refuses a comma", comma.name == "ValueError", f"{comma.name}: {comma.message}")
kit.table(["What this notebook established", "The evidence"],
          [("A notebook is true only if it runs from a fresh kernel, top to bottom",
            "NameError for a name no cell had made"),
           ("One order is a dictionary; the file is a list of them", "ORDERS[0]['amount'] is 2300"),
           ("Counting is start, update, finish", "30 orders, and a desk check of the first five passes"),
           ("A trace is read from its last line up", "TypeError, +=, 'int' and 'str', line 3"),
           ("int() is a patch with a cost", "Rs 5,44,810 today; a comma breaks it")],
          caption="Summary: the first count")
kit.driver_tree({"label": "revenue", "note": "Rs 5,44,810, counted", "kind": "known", "children": [
    {"label": "customers", "note": "notebook 3"},
    {"label": "revenue per customer", "note": "notebook 3", "children": [
        {"label": "orders per customer", "note": "notebook 3"},
        {"label": "revenue per order", "note": "notebook 4"}]}]},
    title="Where the tree stands: one leaf counted, the rest next")
````
