## Step 1. Read it before you run it

The loop below totals the delivered orders and one line sits in the wrong place. Decide what it will print before you run it.

````python
kit.flow(["read the broken loop", "predict the wrong number", "repair it", "prove the repair"], lit=0)
````

````python
# TODO 1. What will that loop print?
#   a) 25720, the correct delivered total
#   b) 0, because the reset wipes it at the end
#   c) 1460, the last delivered order's amount
#   d) 58210, every order in the file
total = 0
for r in records:
    if r["status"] == "delivered":
        total = 0
        total = total + int(r["amount"])

print(total)

prediction = 1460
````

````python
kit.check("the loop printed the last delivered order rather than the total", total == 1460,
          f"it printed {total}")
kit.check("your prediction matched what happened", prediction == total,
          f"you predicted {prediction} and it printed {total}")
````

Option a is the number the loop is meant to produce and never reaches, because the reset runs on every qualifying record. Option b would be right if the reset ran after the addition rather than before it. Option d is the total of all thirty amounts, which no condition in this loop selects.

## Step 3. Repair it

Move the accumulator's starting value to where the loop cannot reach it, and run the same walk again.

````python
kit.flow(["read the broken loop", "predict the wrong number", "repair it", "prove the repair"], lit=2)
````

````python
# TODO 3. What belongs above the loop?
#   a) total = 0
#   b) total = total
#   c) del total
#   d) total = records
total = 0
for r in records:
    if r["status"] == "delivered":
        total = total + int(r["amount"])

print(total)
````

````python
kit.check("the repaired loop totals Rs 25,720", total == 25720, f"total is {total}")
kit.check("the repair is seventeen times the broken answer", total // 1460 == 17)
````

Option b leaves `total` holding whatever the previous cell left on the bench, so the answer depends on what you ran before. Options c and d do not give the accumulator a number to start from, so the first addition raises.

## Step 4. Prove the repair on a second condition

A repair that works once may still be luck. Run the same shape on the returned orders and confirm both numbers at the same time.

````python
kit.flow(["read the broken loop", "predict the wrong number", "repair it", "prove the repair"], lit=3)
````

````python
# TODO 4. What fraction of the two totals is delivered?
#   a) delivered_total / returned_total
#   b) delivered_total / 30
#   c) delivered_total / (delivered_total + returned_total)
#   d) delivered_total - returned_total
delivered_total = 0
returned_total = 0
for r in records:
    if r["status"] == "delivered":
        delivered_total = delivered_total + int(r["amount"])
    elif r["status"] == "returned":
        returned_total = returned_total + int(r["amount"])

print(delivered_total, returned_total)
share = delivered_total / (delivered_total + returned_total)
print(round(share, 3))
````

````python
kit.check("delivered still totals Rs 25,720", delivered_total == 25720)
kit.check("returned totals Rs 13,670", returned_total == 13670)
kit.check("delivered is a share between nought and one", 0 < share < 1, f"share is {share:.3f}")
````
