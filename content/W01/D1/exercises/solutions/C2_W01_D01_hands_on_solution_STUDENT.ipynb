## 2. Revenue

One amount in this file may not be a number. Decide what to do about it before you write the
loop, and write the reason in the comment.

```python
# Reason for how I handle a non-numeric amount: one row stores its amount as text, so int() is
# applied to every amount rather than to that one row, because a rule that fires on one row today
# will miss the second row tomorrow. The decision is written here because on Wednesday it gets a log.
revenue = 0
for order in ORDERS:
    revenue = revenue + int(order["amount"])
print(f"revenue: Rs {revenue:,}")
```

```python
kit.check("revenue is a whole number of rupees", isinstance(revenue, int), f"got {type(revenue).__name__}")
kit.check("revenue is above Rs 5 lakh", revenue > 500000, f"got Rs {revenue:,}")
```

## 3. Customers, and the rate that needs its denominator

```python
customers = set()
for order in ORDERS:
    customers.add(order["customer_id"])

orders_per_customer = len(ORDERS) / len(customers)
print(f"customers: {len(customers)}")
print(f"orders per customer: {orders_per_customer:.2f}")
```

```python
kit.check("every customer counted once", len(customers) < len(ORDERS),
          f"{len(customers)} customers against {len(ORDERS)} orders")
kit.check("orders per customer is at least 1", orders_per_customer >= 1,
          f"got {orders_per_customer:.2f}")
```

## 4. The median, by hand

Sort the amounts, then take the middle one. With an even number of orders, take the average of
the two in the middle.

```python
amounts = sorted(int(o["amount"]) for o in ORDERS)
middle = len(amounts) // 2
median = amounts[middle] if len(amounts) % 2 else (amounts[middle - 1] + amounts[middle]) / 2
mean = sum(amounts) / len(amounts)
print(f"mean:   Rs {mean:,.0f}")
print(f"median: Rs {median:,.0f}")
```

```python
kit.check("the median sits inside the ordinary range", 500 < median < 6000, f"Rs {median:,.0f}")
kit.check("the mean is well above the median", mean > 4 * median,
          f"mean Rs {mean:,.0f} against median Rs {median:,.0f}")
```
